# PSF Capture Workflow

## Overview
Automated PSF capture with SLM phase patterns and Z-axis scanning.

## Workflow
1. **Setup** - Hardware connection, Fresnel test, file selection, scan parameters
2. **Capture** - Automated Z-scan with progress tracking
3. **Organize** - Sort TIFF files by pattern name

---

# Step 1: Capture Setup

**Features:**
- Connect to hardware (SLM, Z-stage, rotation stage)
- Upload Fresnel test pattern to verify system
- Set rotation stage angle (based on M value)
- Select phase pattern files for capture
- View scan statistics before capture

```python
from capture_gui import create_capture_setup_gui

# ============================================================
# Rotation Angle Presets (modify as needed)
# ============================================================
# Standard patterns (no randomness)
ROTATION_ANGLES = {
    'M3': 285,
    'M5': 270,
    'M7': 264,
    'M9': 261,
}

# Patterns with randomness (typically need different angles)
ROTATION_ANGLES_RAND = {
    'M3': 285,
    'M5': 285,
    'M7': 270,
    'M9': 264,
}

setup_gui = create_capture_setup_gui(
    output_dir='./output',
    rotation_angles=ROTATION_ANGLES,
    rotation_angles_rand=ROTATION_ANGLES_RAND
)
```

# Step 2: Automated Capture

**Features:**
- Automatic Z-scan with camera triggering
- Real-time progress bar and ETA
- Auto rotation stage switching when M value changes

**Before running:**
- Ensure camera software is open and ready
- First run will prompt to click camera trigger button

```python
from capture_gui import create_auto_capture_gui
from nas_mapper import quick_map

# Map NAS drive (if saving to network location)
success, msg = quick_map()
if not success:
    print(f"Warning: NAS mapping failed: {msg}")

# Create capture GUI
# - Set save directory and filename prefix
# - Click "Start Capture" to begin
# - Progress bar shows current frame and ETA

capture_gui = create_auto_capture_gui(
    setup_gui,
    save_dir=r"Z:\\SLM_super_resolution\\data\\for_auto_scan\\"
)
```

# Step 3: Organize Data

Organize captured TIFF files into folders by pattern name.

```python
# Organize captured data into folders
# Uses the prefix from capture GUI

capture_gui.organize_data()
```
