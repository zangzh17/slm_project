# Phase Optimizer Quick Test

This notebook provides a GUI-based interface for quick testing of phase optimization parameters.

**Features:**
- Full GUI control of all optimization parameters
- Immediate optimization with one click
- Real-time visualization and analysis
- **No file saving** - purely for testing and experimentation

**New Features:**
- `randomness` parameter for PSF position randomization
- `random_seed` for reproducible random patterns

```python
# Setup and imports
import torch
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib
matplotlib.use('agg')  # Use non-interactive backend to avoid duplicate figures
import matplotlib.pyplot as plt

import config
from phase_generators import PhaseGenerator
from optics_utils import compute_psf_centers
from visualization import plot_phase, plot_2d_comparisons, plot_energy_distribution

# Prevent duplicate widget output in VS Code
%matplotlib inline

# Select device
device = torch.device('cuda') if torch.cuda.is_available() else torch.device('cpu')
print(f"Using device: {device}")
```

```python
class QuickTestGUI:
    """
    GUI for quick testing of phase optimization with all parameters.
    Supports immediate optimization without saving results.
    """
    
    def __init__(self, device):
        self.device = device
        self.optimizer = None
        self.phase_8bit = None
        
        self._create_widgets()
        self._setup_layout()
        self._setup_callbacks()
    
    def _create_widgets(self):
        """Create all GUI widgets."""
        style = {'description_width': '120px'}
        layout_narrow = widgets.Layout(width='220px')
        layout_wide = widgets.Layout(width='280px')
        
        # === Basic Parameters ===
        self.w_M = widgets.IntSlider(
            value=5, min=2, max=15, step=1,
            description='M (array size):',
            style=style, layout=layout_wide
        )
        
        self.w_N = widgets.IntSlider(
            value=256, min=64, max=1152, step=32,
            description='N (pixels):',
            style=style, layout=layout_wide
        )
        
        self.w_focal_length = widgets.FloatText(
            value=73.9, description='Focal (mm):',
            style=style, layout=layout_narrow
        )
        
        self.w_two_pi_value = widgets.IntText(
            value=210, description='2π Value:',
            style=style, layout=layout_narrow
        )
        
        # === Mode Selection ===
        self.w_mode = widgets.RadioButtons(
            options=['fresnel', 'optimized'],
            value='optimized',
            description='Mode:',
            style=style
        )
        
        # === Optimization Parameters ===
        self.w_overlap_ratio = widgets.FloatSlider(
            value=0.3, min=0.0, max=0.9, step=0.05,
            description='Overlap Ratio:',
            style=style, layout=layout_wide,
            readout_format='.2f'
        )
        
        self.w_airy_correction = widgets.FloatSlider(
            value=1.0, min=0.1, max=2.0, step=0.1,
            description='Airy Correction:',
            style=style, layout=layout_wide,
            readout_format='.1f'
        )
        
        self.w_center_blend = widgets.FloatSlider(
            value=0.0, min=0.0, max=1.0, step=0.1,
            description='Center Blend:',
            style=style, layout=layout_wide,
            readout_format='.1f'
        )
        
        # === Randomness Parameters (NEW) ===
        self.w_randomness = widgets.FloatSlider(
            value=0.0, min=0.0, max=0.5, step=0.05,
            description='Randomness:',
            style=style, layout=layout_wide,
            readout_format='.2f'
        )
        
        self.w_random_seed = widgets.IntText(
            value=42, description='Random Seed:',
            style=style, layout=layout_narrow
        )
        
        self.w_use_seed = widgets.Checkbox(
            value=True, description='Use Seed',
            style=style
        )
        
        # === Depth Parameters ===
        self.w_depth_in_focus = widgets.Textarea(
            value='-0.5, 0.5',
            placeholder='e.g., -0.5, 0.5',
            description='Depth in Focus:',
            style=style,
            layout=widgets.Layout(width='300px', height='50px')
        )
        
        self.w_dof_correction = widgets.FloatSlider(
            value=1.0, min=0.1, max=3.0, step=0.1,
            description='DOF Correction:',
            style=style, layout=layout_wide,
            readout_format='.1f'
        )
        
        # === Training Parameters ===
        self.w_lr = widgets.FloatLogSlider(
            value=0.5, base=10, min=-2, max=0, step=0.1,
            description='Learning Rate:',
            style=style, layout=layout_wide,
            readout_format='.3f'
        )
        
        self.w_ni = widgets.IntSlider(
            value=100, min=10, max=1000, step=10,
            description='Iterations:',
            style=style, layout=layout_wide
        )
        
        self.w_upsampling = widgets.FloatSlider(
            value=2.0, min=1.0, max=4.0, step=0.5,
            description='Upsampling:',
            style=style, layout=layout_wide,
            readout_format='.1f'
        )
        
        # === Loss Weights ===
        self.w_weight_mse = widgets.FloatText(
            value=1.0, description='MSE:',
            style={'description_width': '80px'},
            layout=widgets.Layout(width='150px')
        )
        
        self.w_weight_depth = widgets.FloatText(
            value=1.0, description='Depth:',
            style={'description_width': '80px'},
            layout=widgets.Layout(width='150px')
        )
        
        self.w_weight_eff_mean = widgets.FloatText(
            value=20.0, description='Eff Mean:',
            style={'description_width': '80px'},
            layout=widgets.Layout(width='150px')
        )
        
        self.w_weight_eff_std = widgets.FloatText(
            value=50.0, description='Eff Std:',
            style={'description_width': '80px'},
            layout=widgets.Layout(width='150px')
        )
        
        # === Mask Parameters ===
        self.w_mask_count = widgets.IntSlider(
            value=0, min=0, max=9, step=1,
            description='Mask Count:',
            style=style, layout=layout_wide
        )
        
        self.w_interleaving = widgets.Dropdown(
            options=['checkerboard', 'coarse1', 'coarse2', 'coarse3'],
            value='coarse1',
            description='Interleaving:',
            style=style, layout=layout_narrow
        )
        
        # === Action Buttons ===
        self.w_run_btn = widgets.Button(
            description='Run Optimization',
            button_style='success',
            layout=widgets.Layout(width='200px', height='40px')
        )
        
        self.w_visualize_btn = widgets.Button(
            description='Visualize Results',
            button_style='info',
            layout=widgets.Layout(width='200px', height='40px')
        )
        
        self.w_analyze_btn = widgets.Button(
            description='Efficiency Analysis',
            button_style='primary',
            layout=widgets.Layout(width='200px', height='40px')
        )
        
        self.w_show_centers_btn = widgets.Button(
            description='Show PSF Centers',
            button_style='warning',
            layout=widgets.Layout(width='200px', height='40px')
        )
        
        self.w_psf_slices_btn = widgets.Button(
            description='PSF Slices',
            button_style='',
            layout=widgets.Layout(width='200px', height='40px')
        )
        
        # === Info Display ===
        self.w_info = widgets.HTML(
            value='<i>Configure parameters and click "Run Optimization"</i>',
            layout=widgets.Layout(width='100%')
        )
        
        # === Output Areas ===
        self.w_log_output = widgets.Output(
            layout=widgets.Layout(width='100%', height='200px', overflow='auto', border='1px solid #ccc')
        )
        
        self.w_plot_output = widgets.Output(
            layout=widgets.Layout(width='100%')
        )
    
    def _setup_layout(self):
        """Setup the widget layout."""
        # Basic params section
        basic_section = widgets.VBox([
            widgets.HTML('<h4>Basic Parameters</h4>'),
            self.w_M,
            self.w_N,
            widgets.HBox([self.w_focal_length, self.w_two_pi_value]),
            self.w_mode
        ])
        
        # Optimization params section
        opt_section = widgets.VBox([
            widgets.HTML('<h4>Optimization Parameters</h4>'),
            self.w_overlap_ratio,
            self.w_airy_correction,
            self.w_center_blend,
            widgets.HTML('<b>Randomness (PSF position jitter):</b>'),
            self.w_randomness,
            widgets.HBox([self.w_random_seed, self.w_use_seed])
        ])
        
        # Depth params section
        depth_section = widgets.VBox([
            widgets.HTML('<h4>Depth Parameters</h4>'),
            self.w_depth_in_focus,
            self.w_dof_correction
        ])
        
        # Training params section
        train_section = widgets.VBox([
            widgets.HTML('<h4>Training Parameters</h4>'),
            self.w_lr,
            self.w_ni,
            self.w_upsampling
        ])
        
        # Weights section
        weights_section = widgets.VBox([
            widgets.HTML('<h4>Loss Weights</h4>'),
            widgets.HBox([self.w_weight_mse, self.w_weight_depth]),
            widgets.HBox([self.w_weight_eff_mean, self.w_weight_eff_std])
        ])
        
        # Mask params section
        mask_section = widgets.VBox([
            widgets.HTML('<h4>Mask Parameters</h4>'),
            self.w_mask_count,
            self.w_interleaving
        ])
        
        # Action buttons
        action_section = widgets.VBox([
            widgets.HTML('<h4>Actions</h4>'),
            widgets.HBox([self.w_run_btn, self.w_visualize_btn]),
            widgets.HBox([self.w_analyze_btn, self.w_show_centers_btn]),
            self.w_psf_slices_btn
        ])
        
        # Left column
        left_col = widgets.VBox([
            basic_section,
            opt_section,
            depth_section
        ])
        
        # Right column
        right_col = widgets.VBox([
            train_section,
            weights_section,
            mask_section,
            action_section
        ])
        
        # Main layout
        params_panel = widgets.HBox([left_col, right_col])
        
        self.main_layout = widgets.VBox([
            widgets.HTML('<h2>Phase Optimizer Quick Test</h2>'),
            self.w_info,
            params_panel,
            widgets.HTML('<h4>Log Output</h4>'),
            self.w_log_output,
            widgets.HTML('<h4>Visualization</h4>'),
            self.w_plot_output
        ])
    
    def _setup_callbacks(self):
        """Setup widget callbacks."""
        self.w_run_btn.on_click(self._on_run)
        self.w_visualize_btn.on_click(self._on_visualize)
        self.w_analyze_btn.on_click(self._on_analyze)
        self.w_show_centers_btn.on_click(self._on_show_centers)
        self.w_psf_slices_btn.on_click(self._on_psf_slices)
        
        # Update info on parameter changes
        self.w_overlap_ratio.observe(self._update_info, names='value')
        self.w_M.observe(self._update_info, names='value')
        self.w_N.observe(self._update_info, names='value')
        self.w_randomness.observe(self._update_info, names='value')
    
    def _update_info(self, change=None):
        """Update info display."""
        M = self.w_M.value
        N = self.w_N.value
        overlap = self.w_overlap_ratio.value
        randomness = self.w_randomness.value
        
        # Calculate alpha
        if overlap > 0 and M > 0:
            base_size = N / M
            extended_size = overlap * (N - N / M) + N / M
            alpha = extended_size / base_size
            dof_info = f'alpha={alpha:.2f}, DOF x{1/alpha**2:.2f}, Airy x{1/alpha:.2f}'
        else:
            dof_info = 'Standard Fresnel (no overlap)'
        
        # Randomness info
        if randomness > 0:
            pitch = N / M
            sigma = randomness * pitch
            rand_info = f'sigma = {sigma:.1f} px'
        else:
            rand_info = 'No randomization'
        
        self.w_info.value = f'''
        <div style="background: #e8f4f8; padding: 10px; border-radius: 5px; margin: 5px 0;">
            <b>Current Config:</b> {M}x{M} array, {N}x{N} pixels | 
            <b>Overlap:</b> {dof_info} | 
            <b>Randomness:</b> {rand_info}
        </div>
        '''
    
    def _get_params(self):
        """Get current parameters as a dictionary."""
        # Parse depth list
        try:
            depth_text = self.w_depth_in_focus.value.strip()
            if depth_text:
                depth_list = [float(x.strip()) for x in depth_text.split(',')]
            else:
                depth_list = [-0.5, 0.5]
        except:
            depth_list = [-0.5, 0.5]
        
        N = self.w_N.value
        
        params = {
            'M': self.w_M.value,
            'N': N,
            'output_size': N,  # Keep N == output_size for simplicity
            'focal_length': self.w_focal_length.value * 1e-3,
            'two_pi_value': self.w_two_pi_value.value,
            'overlap_ratio': self.w_overlap_ratio.value,
            'airy_correction': self.w_airy_correction.value,
            'center_blend': self.w_center_blend.value,
            'randomness': self.w_randomness.value,
            'random_seed': self.w_random_seed.value if self.w_use_seed.value else None,
            'depth_in_focus': depth_list,
            'depth_out_focus': None,
            'dof_correction': self.w_dof_correction.value,
            'lr': self.w_lr.value,
            'ni': self.w_ni.value,
            'show_iters': max(10, self.w_ni.value // 10),
            'mask_count': self.w_mask_count.value,
            'interleaving': self.w_interleaving.value,
            'masked_airy_correction': 1.0,
            'focusing_eff_correction': 1.0,
            'psf_energy_level': 1.0,
            'weights': {
                'mse': self.w_weight_mse.value,
                'depth_in_focus': self.w_weight_depth.value,
                'depth_out_focus': 0.0,
                'masked': 1.0 if self.w_mask_count.value > 0 else 0.0,
                'eff_mean': self.w_weight_eff_mean.value,
                'eff_std': self.w_weight_eff_std.value
            },
            'shape': [1152, 1920],
            'roi_center_x': 960,
            'roi_center_y': 576
        }
        return params
    
    def _on_run(self, btn):
        """Run optimization."""
        with self.w_log_output:
            clear_output()
            print("="*60)
            print("Starting optimization...")
            print("="*60)
            
            params = self._get_params()
            mode = self.w_mode.value
            
            print(f"Mode: {mode}")
            print(f"M: {params['M']}, N: {params['N']}")
            print(f"Randomness: {params['randomness']}, Seed: {params['random_seed']}")
            print()
            
            try:
                self.optimizer = PhaseGenerator(params, device=self.device, mode=mode)
                
                if mode == 'fresnel':
                    self.optimizer.generate(mode='fresnel')
                else:
                    self.optimizer.generate(
                        mode='optimized',
                        upsampling=self.w_upsampling.value,
                        visualize=False
                    )
                
                self.phase_8bit = self.optimizer.update_phase_8bit()
                
                print()
                print("="*60)
                print("Optimization completed!")
                print(f"Phase shape: {self.phase_8bit.shape}")
                print("="*60)
                
            except Exception as e:
                print(f"Error: {e}")
                import traceback
                traceback.print_exc()
        
        self._update_info()
    
    def _on_visualize(self, btn):
        """Visualize basic results."""
        if self.optimizer is None:
            with self.w_log_output:
                print("Please run optimization first!")
            return
        
        with self.w_plot_output:
            clear_output(wait=True)
            
            try:
                # Close all existing figures to ensure fresh plots
                plt.close('all')
                
                plot_phase(self.optimizer)
                plt.show()
                
                plot_2d_comparisons(self.optimizer)
                plt.show()
                
                print("Visualization completed!")
            except Exception as e:
                print(f"Visualization error: {e}")
                import traceback
                traceback.print_exc()
    
    def _on_analyze(self, btn):
        """Efficiency analysis (without 1D cross-sections)."""
        if self.optimizer is None:
            with self.w_log_output:
                print("Please run optimization first!")
            return
        
        with self.w_plot_output:
            clear_output(wait=True)
            
            try:
                # Close all existing figures to ensure fresh plots
                plt.close('all')
                
                # Only show efficiency analysis (removed plot_cross_sections)
                plot_energy_distribution(self.optimizer, upsampling=3)
                plt.show()
                
                # Print loss history if available
                if self.optimizer.history:
                    print("\n" + "="*40)
                    print("Loss History Summary")
                    print("="*40)
                    for key, values in self.optimizer.history.items():
                        if values:
                            print(f"{key}: {values[0]:.4f} -> {values[-1]:.4f}")
                
                print("\nAnalysis completed!")
                            
            except Exception as e:
                print(f"Analysis error: {e}")
                import traceback
                traceback.print_exc()
    
    def _on_show_centers(self, btn):
        """Show PSF centers visualization."""
        with self.w_plot_output:
            clear_output(wait=True)
            
            # Close all existing figures
            plt.close('all')
            
            params = self._get_params()
            M = params['M']
            N = params['N']
            
            fig, axes = plt.subplots(1, 2, figsize=(14, 6))
            
            # Left: No randomness (reference)
            result_ref = compute_psf_centers(
                M=M, overlap_ratio=params['overlap_ratio'],
                center_blend=params['center_blend'], z_ratio=1.0,
                N=N, output_size=N, device=self.device,
                randomness=0.0
            )
            centers_ref = result_ref['centers_pixel'].cpu().numpy()
            
            ax = axes[0]
            ax.scatter(centers_ref[:, 1], centers_ref[:, 0], c='blue', s=100, label='PSF Centers')
            ax.set_xlim(0, N)
            ax.set_ylim(N, 0)
            ax.set_aspect('equal')
            ax.set_xlabel('X (pixels)')
            ax.set_ylabel('Y (pixels)')
            ax.set_title(f'Reference (No Randomness) - {N}x{N}')
            ax.grid(True, alpha=0.3)
            ax.legend()
            
            # Right: With current randomness
            result_rand = compute_psf_centers(
                M=M, overlap_ratio=params['overlap_ratio'],
                center_blend=params['center_blend'], z_ratio=1.0,
                N=N, output_size=N, device=self.device,
                randomness=params['randomness'],
                random_seed=params['random_seed']
            )
            centers_rand = result_rand['centers_pixel'].cpu().numpy()
            
            ax = axes[1]
            ax.scatter(centers_ref[:, 1], centers_ref[:, 0], c='lightgray', s=100, marker='x', label='Reference')
            ax.scatter(centers_rand[:, 1], centers_rand[:, 0], c='red', s=100, label='Randomized')
            
            # Draw arrows showing displacement
            if params['randomness'] > 0:
                for i in range(len(centers_ref)):
                    ax.annotate('', xy=(centers_rand[i, 1], centers_rand[i, 0]),
                               xytext=(centers_ref[i, 1], centers_ref[i, 0]),
                               arrowprops=dict(arrowstyle='->', color='green', alpha=0.5))
            
            ax.set_xlim(0, N)
            ax.set_ylim(N, 0)
            ax.set_aspect('equal')
            ax.set_xlabel('X (pixels)')
            ax.set_ylabel('Y (pixels)')
            ax.set_title(f'Randomness = {params["randomness"]}, Seed = {params["random_seed"]}')
            ax.grid(True, alpha=0.3)
            ax.legend()
            
            plt.tight_layout()
            plt.show()
            
            # Print statistics
            if params['randomness'] > 0 and 'random_offsets' in result_rand:
                offsets = result_rand['random_offsets'].cpu().numpy()
                print(f"\nOffset Statistics:")
                print(f"  Mean offset: ({offsets[:, 0].mean():.2f}, {offsets[:, 1].mean():.2f}) pixels")
                print(f"  Std offset: ({offsets[:, 0].std():.2f}, {offsets[:, 1].std():.2f}) pixels")
                print(f"  Max offset: {np.abs(offsets).max():.2f} pixels")
            
            print("\nPSF centers visualization completed!")
    
    def _on_psf_slices(self, btn):
        """Show individual PSF slices with template overlay and group navigation."""
        if self.optimizer is None:
            with self.w_log_output:
                print("Please run optimization first!")
            return
        
        with self.w_plot_output:
            clear_output(wait=True)
            
            # Get parameters
            params = self._get_params()
            M = params['M']
            N = params['N']
            upsampling = 2.0
            N_up = int(N * upsampling)
            
            # Get PSF centers with randomness
            randomness = getattr(self.optimizer, 'randomness', 0.0)
            random_seed = getattr(self.optimizer, 'random_seed', None)
            
            center_info = compute_psf_centers(
                M=M, overlap_ratio=params['overlap_ratio'],
                center_blend=params['center_blend'], z_ratio=1.0,
                N=N_up, output_size=N_up, device=self.device,
                randomness=randomness,
                random_seed=random_seed
            )
            centers = center_info['centers_pixel'].cpu().numpy()
            
            # Get intensity
            with torch.no_grad():
                I_opt = self.optimizer.forward(upsampling=upsampling).cpu().numpy()
            
            # Get template if available
            template = None
            if hasattr(self.optimizer, 'template') and self.optimizer.template is not None:
                t = self.optimizer.template
                if t.shape[0] == N_up and t.shape[1] == N_up:
                    template = t.cpu().numpy()
            
            # Calculate slice parameters
            pitch = N_up / M
            slice_half_width = int(pitch * 0.5)  # 50% of pitch on each side
            
            num_psfs = M * M
            psfs_per_group = 5
            num_groups = (num_psfs + psfs_per_group - 1) // psfs_per_group
            
            # Extract slices for all PSFs
            psf_slices = []
            for i in range(num_psfs):
                cy, cx = int(centers[i, 0]), int(centers[i, 1])
                
                # Define slice region
                x_start = max(0, cx - slice_half_width)
                x_end = min(N_up, cx + slice_half_width)
                
                # Extract horizontal slice through PSF center
                if 0 <= cy < N_up:
                    slice_data = I_opt[cy, x_start:x_end]
                    # Normalize
                    if slice_data.max() > 0:
                        slice_data = slice_data / slice_data.max()
                    
                    # Extract template slice if available
                    template_slice = None
                    if template is not None and 0 <= cy < template.shape[0]:
                        template_slice = template[cy, x_start:x_end]
                        if template_slice.max() > 0:
                            template_slice = template_slice / template_slice.max()
                else:
                    slice_data = np.zeros(x_end - x_start)
                    template_slice = None
                
                psf_slices.append({
                    'data': slice_data,
                    'template': template_slice,
                    'center': (cy, cx),
                    'x_range': (x_start, x_end),
                    'index': i,
                    'row': i // M,
                    'col': i % M
                })
            
            # Create interactive widget for group selection
            def plot_group(group_idx):
                plt.close('all')
                
                start_idx = group_idx * psfs_per_group
                end_idx = min(start_idx + psfs_per_group, num_psfs)
                group_psfs = psf_slices[start_idx:end_idx]
                
                n_cols = len(group_psfs)
                fig, axes = plt.subplots(2, n_cols, figsize=(4 * n_cols, 6))
                
                if n_cols == 1:
                    axes = axes.reshape(-1, 1)
                
                for j, psf_info in enumerate(group_psfs):
                    # Top row: Linear scale
                    ax = axes[0, j]
                    x_pixels = np.arange(len(psf_info['data']))
                    
                    # Plot template first (if available) as background
                    if psf_info['template'] is not None:
                        ax.fill_between(x_pixels, psf_info['template'], alpha=0.3, color='orange', label='Template')
                        ax.plot(x_pixels, psf_info['template'], 'orange', linewidth=1, alpha=0.7)
                    
                    # Plot actual PSF
                    ax.plot(x_pixels, psf_info['data'], 'b-', linewidth=1.5, label='Optimized')
                    ax.fill_between(x_pixels, psf_info['data'], alpha=0.3, color='blue')
                    
                    ax.set_title(f"PSF #{psf_info['index']} [{psf_info['row']},{psf_info['col']}]", fontsize=10)
                    ax.set_xlabel('Pixels')
                    ax.set_ylabel('Intensity (norm)')
                    ax.set_ylim(0, 1.1)
                    ax.grid(True, alpha=0.3)
                    if j == 0:
                        ax.legend(fontsize=8)
                    
                    # Bottom row: Log scale
                    ax = axes[1, j]
                    
                    if psf_info['template'] is not None:
                        ax.semilogy(x_pixels, psf_info['template'] + 1e-6, 'orange', linewidth=1, alpha=0.7, label='Template')
                    
                    ax.semilogy(x_pixels, psf_info['data'] + 1e-6, 'b-', linewidth=1.5, label='Optimized')
                    ax.set_xlabel('Pixels')
                    ax.set_ylabel('Intensity (log)')
                    ax.set_ylim(1e-4, 2)
                    ax.grid(True, alpha=0.3)
                
                fig.suptitle(f'PSF Horizontal Slices - Group {group_idx + 1}/{num_groups} '
                            f'(PSFs {start_idx}-{end_idx-1})', fontsize=12, fontweight='bold')
                plt.tight_layout()
                plt.show()
                
                # Print info
                print(f"\nGroup {group_idx + 1}/{num_groups}: PSFs #{start_idx} to #{end_idx-1}")
                print(f"Slice width: {2 * slice_half_width} pixels (50% of pitch = {pitch:.1f} px)")
                if randomness > 0:
                    print(f"Randomness: {randomness}, Seed: {random_seed}")
                if template is not None:
                    print("Template overlay: Enabled (orange)")
            
            # Create group selector
            group_selector = widgets.IntSlider(
                value=0, min=0, max=num_groups - 1, step=1,
                description='Group:',
                continuous_update=False,
                layout=widgets.Layout(width='400px')
            )
            
            # Navigation buttons
            prev_btn = widgets.Button(description='< Prev', layout=widgets.Layout(width='100px'))
            next_btn = widgets.Button(description='Next >', layout=widgets.Layout(width='100px'))
            
            def on_prev(b):
                if group_selector.value > 0:
                    group_selector.value -= 1
            
            def on_next(b):
                if group_selector.value < num_groups - 1:
                    group_selector.value += 1
            
            prev_btn.on_click(on_prev)
            next_btn.on_click(on_next)
            
            # Plot output for interactive updates
            plot_out = widgets.Output()
            
            def update_plot(change):
                with plot_out:
                    clear_output(wait=True)
                    plot_group(group_selector.value)
            
            group_selector.observe(update_plot, names='value')
            
            # Display controls
            controls = widgets.HBox([prev_btn, group_selector, next_btn])
            display(widgets.HTML(f'<h4>Individual PSF Slices ({num_psfs} PSFs in {num_groups} groups)</h4>'))
            display(controls)
            display(plot_out)
            
            # Initial plot
            with plot_out:
                plot_group(0)
    
    def display(self):
        """Display the GUI."""
        self._update_info()
        display(self.main_layout)
```

```python
# Launch the GUI
gui = QuickTestGUI(device=device)
gui.display()
```

## Usage Notes

### Quick Start
1. Adjust parameters using the sliders and input fields
2. Click **"Run Optimization"** to execute
3. Click **"Visualize Results"** to see phase and PSF patterns
4. Click **"Detailed Analysis"** for cross-sections and efficiency plots
5. Click **"Show PSF Centers"** to visualize the effect of randomness

### Key Parameters

| Parameter | Description |
|-----------|-------------|
| **M** | Microlens array size (M×M) |
| **N** | ROI size in pixels (use smaller values like 256 for faster testing) |
| **Overlap Ratio** | Adjacent lens overlap (0=standard Fresnel, >0=super-resolution) |
| **Randomness** | PSF position jitter (0=periodic, >0=random offset with σ=randomness×pitch) |
| **Random Seed** | For reproducible random patterns |
| **Depth in Focus** | DOF units for extended depth of focus |

### Testing Tips
- Use **N=256** and **ni=100** for quick tests
- Use **lr=0.5** for faster convergence in testing
- Compare **randomness=0** vs **randomness=0.1~0.3** to see the effect
- Use the same **random_seed** to reproduce results
